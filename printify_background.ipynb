```python
#@title 🐺 Orioninsist | Profesyonel Arka Plan Temizleme Aracı
#@markdown Bu araç, tişört tasarımlarınızın veya manken görsellerinizin arka planını yapay zeka ile şeffaflaştırır.
#@markdown Form üzerindeki ayarları değiştirip hücreyi çalıştırabilirsiniz. Kodları gizlemek için hücre başlığına çift tıklayabilirsiniz.
#@markdown ---

model_tipi = "isnet-general-use" #@param ["isnet-general-use", "u2net", "u2net_human"]
#@markdown *   **isnet-general-use**: Tişört tasarımları, logolar ve grafikler için en keskin kenar temizleme modelidir.
#@markdown *   **u2net_human**: İnsan model görselleri (Freya) izole edilirken tercih edilmelidir.
#@markdown *   **u2net**: Genel amaçlı arka plan temizleme modelidir.

hassas_kenar_bulma = True #@param {type:"boolean"}
#@markdown *   *İpucu:* Saç veya tüy gibi detaylı kenarları daha hassas temizlemek için aktif bırakın.

arka_plan_esigi = 20 #@param {type:"slider", min:0, max:255, step:5}
#@markdown *   *İpucu:* Arka planın silinme hassasiyetidir. Değer düştükçe daha fazla arka plan silinir.

on_plan_esigi = 270 #@param {type:"slider", min:0, max:300, step:10}
#@markdown *   *İpucu:* Ön plandaki nesnenin korunma eşiğidir. Değer yükseldikçe tasarımın aslı daha çok korunur.

# 1. Gerekli kütüphaneyi kuruyoruz
print("Arka plan temizleme kütüphanesi (rembg) kuruluyor...")
!pip install rembg[gpu] # Colab Pro GPU'nuzdan faydalanmak için [gpu] ekliyoruz

from rembg import remove
from PIL import Image
from google.colab import files
import io
import os

print("\n✅ Kurulum tamamlandı!")

# 2. Görsel yükleme
print("\nLütfen arka planını temizlemek istediğiniz YÜKSEK ÇÖZÜNÜRLÜKLÜ görseli yükleyin:")
uploaded = files.upload()

if uploaded:
    filename = next(iter(uploaded))
    print(f"\n'{filename}' dosyası işleniyor...")

    input_image = Image.open(io.BytesIO(uploaded[filename]))

    # 3. Arka planı temizleme işlemini yap
    print(f"Yapay zeka arka planı temizliyor (Model: {model_tipi}), lütfen bekleyin...")
    output_image = remove(
        input_image, 
        model=model_tipi, 
        alpha_matting=hassas_kenar_bulma, 
        alpha_matting_foreground_threshold=on_plan_esigi, 
        alpha_matting_background_threshold=arka_plan_esigi
    )

    # 4. Sonucu şeffaf PNG olarak kaydet ve indir
    output_filename = f"{os.path.splitext(filename)[0]}_şeffaf_PYTHON.png"
    output_image.save(output_filename)

    print(f"✅ İşlem tamamlandı! Şeffaf arka planlı görseliniz '{output_filename}' olarak kaydedildi.")
    files.download(output_filename)
else:
    print("Hiçbir dosya yüklenmedi.")
```
